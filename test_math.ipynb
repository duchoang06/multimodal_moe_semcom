```python
import cvxpy as cp
import numpy as np

a = cp.Variable((3, 1), nonneg=True)
b = cp.Variable((3, 1), nonneg=True)

constraints = [a >= 0, b >= 0, cp.sum(a) == 1, cp.sum(b) == 1, a <= 1, b <= 1]
objective = cp.Minimize(a.T * b)

prob = cp.Problem(objective, constraints)
prob.solve(verbose=True, solver=cp.MOSEK)

print("Optimal value:", prob.value)
print("Optimal a:", a.value)
print("Optimal b:", b.value)
```
